```python
# alternating x and y direction layers  with propagation distance to detector differnt than that of the layers

class DNN_model_imaging(pl.LightningModule):
    def __init__(
        self, Nx, Ny, Dx, Dy, wl, propag_z, propag_z_detector, N_layers, init="rnd", learning_rate=0.001
    ):
        super().__init__()
        self.learning_rate = learning_rate
        self.wavelength = wl
        self.N_layers = N_layers

        # Trainable 1d phase layers (all same dim.)
        learnedphase_lay_x = []
        learnedphase_lay_y = []
        for i in range(self.N_layers):
            learnedphase_lay_x.append(
                LearnablePhaseLayer1d(Nx, Ny, Dx, Dy, learnable_axis="x", init=init)
            )
            learnedphase_lay_y.append(
                LearnablePhaseLayer1d(Nx, Ny, Dx, Dy, learnable_axis="y", init=init)
            )
        self.phaselayer_x_list = torch.nn.ModuleList(learnedphase_lay_x)
        self.phaselayer_y_list = torch.nn.ModuleList(learnedphase_lay_y)

        # Free space propagation segments, here: all same dist. except for the detector
        self.propalayer_in = PropagationLayer(Nx, Ny, Dx, Dy, wl, propag_z)
        self.propalayer_x = PropagationLayer(Nx, Ny, Dx, Dy, wl, propag_z)
        self.propalayer_y = PropagationLayer(Nx, Ny, Dx, Dy, wl, propag_z)
        # Different propagation distance to the detector
        self.propalayer_out = PropagationLayer(Nx, Ny, Dx, Dy, wl, propag_z_detector)

    def forward(self, x):
        # convert scalar 2-channel input to complex for internal calc.
        x = x[:, 0, ...] + 1j * x[:, 1, ...]

        # propagation to first layer
        x = self.propalayer_in(x)

        # alternate several x/y learnable 1d phases
        for i in range(self.N_layers):
            # learnable x-grating + propagation
            x = self.phaselayer_x_list[i](x)
            x = self.propalayer_x(x)

            # learnable y-grating + propagation
            x = self.phaselayer_y_list[i](x)
            x = self.propalayer_y(x)

        # propagation to detector with different distance
        x = self.propalayer_out(x)

        # convert complex back to separate real / imag channels
        x = torch.stack([x.real, x.imag], axis=1)
        return x
    
    def intensity_mse_loss(self, y_pred, y_true):
        # convert to complex
        y_pred_cplx = y_pred[:, 0, ...] + 1j * y_pred[:, 1, ...]
        y_true_cplx = y_true[:, 0, ...] + 1j * y_true[:, 1, ...]
        
        # intensity
        I_pred = torch.abs(y_pred_cplx)**2
        I_true = torch.abs(y_true_cplx)**2
        
        # scale the loss (not needed, the loss is just very small because of the many zeros)
        return 1e3*torch.nn.functional.mse_loss(I_pred, I_true)

    def configure_optimizers(self):
        optimizer = torch.optim.Adam(self.parameters(), lr=self.learning_rate)

        # lr_scheduler = None
        # scheduler = torch.optim.lr_scheduler.ExponentialLR(optimizer, gamma=0.99)
        scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
            optimizer, patience=5, factor=0.5
        )
        return {
            "optimizer": optimizer,
            "lr_scheduler": {
                "scheduler": scheduler,
                "strict": True,
                "monitor": "train_loss",
                "interval": "step",
                "frequency": 200,
                "name": "learning_rate",
            },
        }

    def training_step(self, train_batch, batch_idx):
        # optimization step
        x, y = train_batch
        y_pred = self.forward(x)
        loss = self.intensity_mse_loss(y_pred, y)

        # logging
        self.log("train_loss", loss, prog_bar=True)
        self.log(
            "learning_rate",
            self.optimizers().optimizer.param_groups[0]["lr"],
            prog_bar=True,
        )

        return loss

    def validation_step(self, val_batch, batch_idx):
        x, y = val_batch
        y_pred = self.forward(x)
        loss = self.intensity_mse_loss(y_pred, y)
        self.log("val_loss", loss, prog_bar=True)

    def predict_step(self, batch, batch_idx, dataloader_idx=0):
        return self(batch[0])
```

```python
# %% main DNN model
class DNN_model_imaging(pl.LightningModule):
    def __init__(
        self, Nx, Ny, Dx, Dy, wl, propag_z, N_layers, init="rnd", learning_rate=0.001
    ):
        super().__init__()
        # pytorch-lightning: configure optimizers in the model class
        self.learning_rate = learning_rate
        self.wavelength = wl
        self.N_layers = N_layers

        # sequence of several trainable phase layers (all same dim.)
        learnedphase_layers = []
        for i in range(self.N_layers):
            learnedphase_layers.append(
                LearnablePhaseLayer2d(Nx, Ny, Dx, Dy, init=init)
            )
        self.phaselayerlist = torch.nn.ModuleList(learnedphase_layers)

        # sequence of several propagation phase layers between, here: all same dist.
        propa_layers = []
        for i in range(self.N_layers + 1):  # one more than learned weights layers
            propa_layers.append(PropagationLayer(Nx, Ny, Dx, Dy, wl, propag_z))
        self.propalayerlist = torch.nn.ModuleList(propa_layers)

    def forward(self, x):
        # convert scalar 2-channel input to complex for internal calc.
        x = x[:, 0, ...] + 1j * x[:, 1, ...]
        x_in = x  # keep original input for later use

        for i in range(self.N_layers):
            x = self.propalayerlist[i](x)

            # nonlinear: apply input modulation after each hidden layer
            # x = x * x_in

            # learnable phase
            x = self.phaselayerlist[i](x)

        # propagation to detector
        x = self.propalayerlist[-1](x)

        # convert complex back to separate real / imag channels
        x = torch.stack([x.real, x.imag], axis=1)
        return x

    def intensity_mse_loss(self, y_pred, y_true):
        # convert to complex
        y_pred_cplx = y_pred[:, 0, ...] + 1j * y_pred[:, 1, ...]
        y_true_cplx = y_true[:, 0, ...] + 1j * y_true[:, 1, ...]
        
        # intensity
        I_pred = torch.abs(y_pred_cplx)**2
        I_true = torch.abs(y_true_cplx)**2
        
        # scale the loss (not needed, the loss is just very small because of the many zeros)
        return 1e3*torch.nn.functional.mse_loss(I_pred, I_true)

    def configure_optimizers(self):
        optimizer = torch.optim.Adam(self.parameters(), lr=self.learning_rate)

        # lr_scheduler = None
        # scheduler = torch.optim.lr_scheduler.ExponentialLR(optimizer, gamma=0.99)
        scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
            optimizer, patience=5, factor=0.5
        )
        return {
            "optimizer": optimizer,
            "lr_scheduler": {
                "scheduler": scheduler,
                "strict": True,
                "monitor": "train_loss",
                "interval": "step",
                "frequency": 200,
                "name": "learning_rate",
            },
        }

    def training_step(self, train_batch, batch_idx):
        # optimization step
        x, y = train_batch
        y_pred = self.forward(x)
        loss = self.intensity_mse_loss(y_pred, y)

        # logging
        self.log("train_loss", loss, prog_bar=True)
        self.log(
            "learning_rate",
            self.optimizers().optimizer.param_groups[0]["lr"],
            prog_bar=True,
        )

        return loss

    def validation_step(self, val_batch, batch_idx):
        x, y = val_batch
        y_pred = self.forward(x)
        loss = self.intensity_mse_loss(y_pred, y)
        self.log("val_loss", loss, prog_bar=True)

    def predict_step(self, batch, batch_idx, dataloader_idx=0):
        return self(batch[0])
```

```python
# %% setup
# nr of layers and propagation distance in between (in meters)
N_layers = 5
#propag_z = 0.1      

# phase modulation layer size (in meters) and layer grid discretization steps
Dx = (50*1e-6)  
#Nx = compute_and_adjust_Nx(Dx)
Nx = 110
Dy = Dx
Ny = Nx
#print(f"Nx (rounded to even): {Nx}")

#Dx = (50) * 1e-6
#Dy =  Dx

# layer grid discretization steps
#Nx = 112
#Ny = Nx

# wavelength (in meters)
wavelength = 633 * 1e-9
do = 1  # diffraction order
propag_z = (Dx*Dx)/(2*do*wavelength)  
propag_z_detector = 0.1 # prop distance to detector in m

# --- init dataset
data_module = ImagingMNISTAmplitudeDataModule(
    r"C:\Users\Loubnan Abou-Hamdan\Desktop\optics w deep learning\1D_models_Peter\1D_Loubnan\odnn\mnist_dataset.h5",
    wavelength=wavelength,
    img_pix_outsize=Nx,
    zoom_factor_out=1,
    batch_size=64,
    # N_load=5000,  # partial data loading for rapid testing
)
data_module.setup()

# get the validation data loader. we'll use it explicitly for testing
val_loader = data_module.val_dataloader()
x_batch, y_batch = next(iter(val_loader))  # testing: get first batch
print("Data: input shape:", x_batch.shape, "output shape:", y_batch.shape)

# --- init DNN model
DNN_model = DNN_model_imaging(
    Nx=Nx,
    Ny=Ny,
    Dx=Dx,
    Dy=Dy,
    wl=wavelength,
    propag_z=propag_z,
    propag_z_detector=propag_z_detector,
    N_layers=N_layers,
    init="zeros",
    learning_rate=0.5,
)

# move the model to GPU (if available)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
DNN_model.to(device)


## %%
# evaluate initialized model
predict_and_plot(
    DNN_model,
    val_loader,
    device,
    plot_idx=14,
    title="init",
    plot='intensity',
)


# %% train the DNN
# plotting callback
class PlottingCallback(pl.callbacks.Callback):
    def __init__(self, val_loader, plot_each_n_epochs=1, plot_each_n_batches=99999):
        super().__init__()
        self.val_loader = val_loader
        self.plot_each_n_batches = plot_each_n_batches
        self.plot_each_n = plot_each_n_epochs

    def on_train_batch_end(self, trainer, pl_module, outputs, batch, batch_idx):
        i_epoch = trainer.current_epoch
        if (
            batch_idx % self.plot_each_n_batches == 0
            and i_epoch % self.plot_each_n == 0
        ):
            predict_and_plot(
                pl_module,
                self.val_loader,
                device,
                plot_idx=74,
                title="epoch: {}, batch {}, val_loss={:.3g}".format(
                    i_epoch, batch_idx, outputs["loss"]
                ),
                savename="plots/01_nonlinearDNN_imaging_phase_e{:0>3}_b{:0>5}.png".format(
                    i_epoch + 1, batch_idx
                ),
                show=False,
                plot='intensity',
            )


trainer = pl.Trainer(
    accelerator=device.type,
    devices=1,
    # profiler="simple",
    max_epochs=10,
    callbacks=[
        # PlottingCallback(val_loader),
        RichProgressBar(),
    ],
)

# the learning rate can be manually set
# DNN_model.learning_rate = 0.1
trainer.fit(DNN_model, data_module)

# %%
predict_and_plot(DNN_model, val_loader, device, plot='intensity')


# %% plot the learned phase maps

plt.figure(figsize=(10, 4))

# plot x layers
diff_l = [_l for _l in DNN_model.phaselayer_x_list]
for i, layer in enumerate(diff_l):
    plt.subplot(2, len(diff_l), i + 1, title=f"phase layer x #{i}")
    phasemap = diff_l[i].phase_weights.detach().cpu().numpy()
    phasemap[phasemap >= np.pi] -= np.pi
    phasemap[phasemap < -np.pi] += np.pi
    plt.plot(phasemap)
    plt.ylim(-1.05 * np.pi, 1.05 * np.pi)

# plot y layers
diff_l = [_l for _l in DNN_model.phaselayer_y_list]
for i, layer in enumerate(diff_l):
    plt.subplot(2, len(diff_l), i + 1 + len(diff_l), title=f"phase layer y #{i}")
    phasemap = diff_l[i].phase_weights.detach().cpu().numpy()
    phasemap[phasemap >= np.pi] -= np.pi
    phasemap[phasemap < -np.pi] += np.pi
    plt.plot(phasemap)
    plt.ylim(-1.05 * np.pi, 1.05 * np.pi)

plt.tight_layout()
plt.show()
```
